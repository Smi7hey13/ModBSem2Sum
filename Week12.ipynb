# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

### Presenting museum visitor data over two weeks using an overlaid line graph offers clear advantages, such as enabling direct day-to-day comparisons and revealing consistent patterns or deviations across weeks. This approach also conserves space, making it ideal for concise presentations. However, it can introduce visual clutter and confusion if the lines intersect frequently or lack clear labeling. In contrast, using separate graphs for each week provides clarity and reduces overlap, allowing viewers to focus on individual trends. Yet, this method makes direct comparisons more difficult and requires more effort and space. Ultimately, the choice depends on whether the goal is detailed weekly analysis or quick cross-week comparison.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = ['2008', '2009', '2010', '2011', '2012', '2013', '2014']
products = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
prices = {
    'Product A': [400, 410, 415, 390, 300, 250, 225],
    'Product B': [400, 410, 415, 390, 300, 250, 225],
    'Product C': [0, 0, 100, 190, 200, 250, 175],
    'Product D': [0, 0, 0, 175, 250, 210, 205],
    'Product E': [0, 0, 0, 0, 0, 100, 210]
}

# Plotting setup
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.1
index = np.arange(len(products))

# Plot bars for each year
for i, year in enumerate(years):
    price_values = [prices[product][i] for product in products]
    ax.bar(index + i * bar_width, price_values, bar_width, label=year)

# Formatting
ax.set_ylabel('Price ($)')
ax.set_title('Average Retail Product Price Per Year')
ax.set_xticks(index + bar_width * (len(years) - 1) / 2)
ax.set_xticklabels(products)
ax.set_ylim(0, 500)
ax.legend(title="Year")
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
